```python
import cv2
import numpy as np
import keras

# Load the pre-trained model
model = keras.models.load_model("asl_classifier.h5")

# Define the labels dictionary
labels_dict = {0: '0', 
               1: 'A', 
               2: 'B', 
               3: 'C', 
               4: 'D', 
               5: 'E',
               6: 'F',
               7: 'G',
               8: 'H',
               9: 'I',
               10: 'J',
               11: 'K',
               12: 'L',
               13: 'M',
               14: 'N',
               15: 'O',
               16: 'P',
               17: 'Q',
               18: 'R',
               19: 'S',
               20: 'T', 
               21: 'U', 
               22: 'V',
               23: 'W',
               24: 'X',
               25: 'Y',
               26: 'Z'}
```

```python
color_dict = (0, 255, 0)
img_size = 128
minValue = 70
string = " "  # Initialize the string to store the results
count = 0

def preprocess_frame(frame):
    # Convert frame to grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    blur = cv2.GaussianBlur(gray, (5, 5), 2)
    th3 = cv2.adaptiveThreshold(blur, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY_INV, 11, 2)
    ret, res = cv2.threshold(th3, minValue, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    resized = cv2.resize(res, (img_size, img_size))
    normalized = resized / 255.0
    reshaped = np.reshape(normalized, (1, img_size, img_size, 1))
    return reshaped

# Initialize webcam
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    
    # Draw a rectangle in the frame (region of interest)
    cv2.rectangle(frame, (50, 50), (178, 178), color_dict, 2)
    
    key = cv2.waitKey(1) & 0xFF
    
    if key == 27:  # ESC key to break
        break
    
    if key == ord('j'):  # 'j' key to capture and predict
        # Extract the region of interest
        roi = frame[50:178, 50:178]
        # Preprocess the frame
        preprocessed_frame = preprocess_frame(roi)
        # Predict the sign language letter
        prediction = model.predict(preprocessed_frame)
        max_index = int(np.argmax(prediction))
        predicted_letter = labels_dict[max_index]
        # Append the predicted letter to the string
        if max_index == 0:
            string += " "
        else:
            string += predicted_letter
    
    # Display the resulting string on the frame
    cv2.putText(frame, string, (275, 50), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (200, 200, 200), 2)
    
    # Display the frame
    cv2.imshow("ASL Sign Detection", frame)

# Release the capture and close windows
cap.release()
cv2.destroyAllWindows()
```
