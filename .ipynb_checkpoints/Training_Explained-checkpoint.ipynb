```python
img_size = 128
# تحديد حجم الصورة إلى 128x128 بكسل
data = []
# إنشاء قائمة لتخزين البيانات (الصور)
target = []
# إنشاء قائمة لتخزين الأهداف (التصنيفات)
c = 0 
# متغير عداد
minValue = 70 
# قيمة العتبة للمعالجة الثنائية
classesf.sort()
# ترتيب العناصر في القائمة classesf بشكل تصاعدي
for category in categories:  # المرور على كل تصنيف في قائمة التصنيفات
    classesf.sort()  
    # ترتيب العناصر في القائمة classesf بشكل تصاعدي
    cat_path = os.path.join(data_path, category)  # إنشاء مسار التصنيف
    print(cat_path)
    # طباعة مسار التصنيف
    classesf.sort() 
    # ترتيب العناصر في القائمة classesf بشكل تصاعدي
    cat_names = os.listdir(cat_path) 
    # الحصول على قائمة بالفئات داخل التصنيف
    classesf.sort()
    # ترتيب العناصر في القائمة classesf بشكل تصاعدي
    print(cat_names)
    # طباعة قائمة الفئات داخل التصنيف
    for classes in cat_names: 
        # المرور على كل فئة داخل التصنيف
        classesf.sort() 
        # ترتيب العناصر في القائمة classesf بشكل تصاعدي
        folder_path = os.path.join(data_path, category, classes) 
        # إنشاء مسار الفئة
        print(folder_path)
        # طباعة مسار الفئة
        classesf.sort()
        # ترتيب العناصر في القائمة classesf بشكل تصاعدي
        img_names = os.listdir(folder_path) 
        # الحصول على قائمة بالصور داخل الفئة
        classesf.sort() 
        # ترتيب العناصر في القائمة classesf بشكل تصاعدي
        #print(img_names)
        for img_name in img_names:
            # المرور على كل صورة داخل الفئة
            #print(img_name)
            classesf.sort() 
            # ترتيب العناصر في القائمة classesf بشكل تصاعدي
            img_path = os.path.join(folder_path, img_name)
            # إنشاء مسار الصورة
            img = cv2.imread(img_path) 
            # قراءة الصورة باستخدام OpenCV

            try:
                gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) 
                # تحويل الصورة إلى تدرجات الرمادي
                blur = cv2.GaussianBlur(gray, (5, 5), 2) 
                # تطبيق فلتر التمويه الغاوسي على الصورة
                th3 = cv2.adaptiveThreshold(blur, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY_INV, 11, 2) 
                # تطبيق العتبة التكيفية على الصورة
                ret, res = cv2.threshold(th3, minValue, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU) 
                # تطبيق العتبة الثنائية مع طريقة Otsu
                #res = np.array(res)
                #print(type(res))
                # تحويل الصورة إلى تدرجات الرمادي
                resized = cv2.resize(res, (img_size, img_size))
                # تغيير حجم الصورة إلى 128x128 بكسل
                # تغيير حجم الصورة إلى 50x50 بكسل، حيث نحتاج إلى حجم ثابت لجميع الصور في مجموعة البيانات
                data.append(resized)
                # إضافة الصورة المعالجة إلى قائمة البيانات
                #print(data)
                target.append(label_classes_dict[classes])
                # إضافة التصنيف المقابل للصورة إلى قائمة الأهداف
            except Exception as e:
                # التعامل مع الأخطاء المحتملة أثناء معالجة الصورة
                print('Exception:', e) 
# طباعة رسالة الخطأ
```

```python
from sklearn.model_selection import train_test_split 
# استيراد الدالة train_test_split لتقسيم البيانات إلى مجموعة تدريب واختبار
train_data, test_data, train_target, test_target = train_test_split(data, new_target, test_size=0.2)
# تقسيم البيانات إلى مجموعة تدريب واختبار بنسبة 20% للاختبار
```

```python
from keras.models import Sequential 
# استيراد النموذج Sequential من مكتبة Keras
#الفئة Sequential في Keras تُستخدم لإنشاء نموذج متسلسل
#وهو نموذج خطي يتكون من تتابع من الطبقات
# يتم إضافة الطبقات الواحدة تلو الأخرى في هذا النوع من النماذج.
from keras.layers import Convolution2D
# استيراد طبقة Convolution2D (التلافيفية) من مكتبة Keras
#طبقة تلافيفية ثنائية الأبعاد: تُستخدم الطبقة Conv2D 
#عدد المرشحات: تحدد عدد المرشحات (filters) التي سيتم استخدامها لاستخراج الميزات.
#حجم المرشح: تحدد أبعاد المرشح (أو النواة)، مثل 3x3 أو 5x5.
#دالة التنشيط: تحدد دالة التنشيط التي سيتم استخدامها بعد تطبيق عملية الالتفاف، مثل ReLU (Rectified Linear Unit).
from keras.layers import MaxPooling2D 
# استيراد طبقة MaxPooling2D (التجمع الأقصى) من مكتبة Keras
#وضيح الطبقة MaxPooling2D:
#التجميع الأعظمي: تقوم هذه الطبقة بتطبيق عملية التجميع الأعظمي (Max Pooling) 
#بمعنى أنها تأخذ النافذة (filter) وتتحرك عبر الصورة وتختار القيمة العظمى في كل نافذة.
#حجم النافذة: تحدد حجم النافذة (أو المصفوفة) التي ستُستخدم لتحديد القيم العظمى. على سبيل المثال، 2x2 أو 3x3.
#الخطوة (Stride): تحدد عدد البكسلات التي تنتقل بها النافذة عند التحرك عبر الصورة.
from keras.layers import Flatten 
# استيراد طبقة Flatten (التسطيح) من مكتبة Keras
#تقوم الطبقة Flatten بأخذ مصفوفة متعددة الأبعاد وتحويلها إلى متجه وحيد البُعد، بدون تغيير في عدد العناصر
from keras.layers import Dense, Dropout 
# استيراد طبقتي Dense (الكثيفة) و Dropout (التسرب) من مكتبة Keras
#تُستخدم طبقة الإسقاطDropout كمبدأ للتنظيم (Regularization) لتقليل مشكلة الإفراط في التكيف (Overfitting)
#طبقة كثيفة (Dense Layer): الطبقة الكثيفة هي طبقة عصبية متصلة بالكامل
# حيث كل وحدة عصبية في هذه الطبقة متصلة بكل وحدة عصبية في الطبقة السابقة.
import os  # استيراد مكتبة os للتعامل مع نظام الملفات
os.environ["CUDA_VISIBLE_DEVICES"] = "1" 
# تعيين الجهاز المرئي لوحدة معالجة الرسوميات (GPU) إلى 1


sz = 128 
# تعيين حجم الصور إلى 128x128 بكسل

# خطوة 1 - بناء الشبكة العصبية الالتفافية (CNN)

# تهيئة الشبكة العصبية
classifier = Sequential()
# إنشاء نموذج Sequential

# إضافة الطبقة الالتفافية الأولى وطبقة التجمع الأقصى
classifier.add(Convolution2D(32, (3, 3), input_shape=(sz, sz, 1), activation='relu'))
# إضافة طبقة Convolution2D مع 32 مرشحًا وحجم مرشح 3x3
classifier.add(MaxPooling2D(pool_size=(2, 2)))
# إضافة طبقة MaxPooling2D بحجم تجمع 2x2

# إضافة الطبقة الالتفافية الثانية وطبقة التجمع الأقصى
classifier.add(Convolution2D(32, (3, 3), activation='relu'))
# إضافة طبقة Convolution2D مع 32 مرشحًا وحجم مرشح 3x3
# input_shape هي الخرائط المميزة من الطبقة الالتفافية السابقة
classifier.add(MaxPooling2D(pool_size=(2, 2))) 
# إضافة طبقة MaxPooling2D بحجم تجمع 2x2
#classifier.add(Convolution2D(32, (3, 3), activation='relu'))
# input_shape هي الخرائط المميزة من الطبقة الالتفافية السابقة
#classifier.add(MaxPooling2D(pool_size=(2, 2)))

# تسطيح الطبقات
classifier.add(Flatten())
# إضافة طبقة Flatten لتحويل الخرائط المميزة إلى شكل مسطح

# إضافة طبقة كثيفة متصلة بالكامل
classifier.add(Dense(units=128, activation='relu')) 
# إضافة طبقة Dense مع 128 وحدة وتفعيل 'relu'
# إضافة طبقة كثيفة مع 128 وحدة وتنشيط ReLU
classifier.add(Dropout(0.40))  # إضافة طبقة Dropout بنسبة 40% لتقليل الإفراط في التعميم
classifier.add(Dense(units=96, activation='relu'))
# إضافة طبقة Dense مع 96 وحدة وتفعيل 'relu'
classifier.add(Dropout(0.40))  # إضافة طبقة Dropout بنسبة 40% لتقليل الإفراط في التعميم
classifier.add(Dense(units=64, activation='relu')) 
# إضافة طبقة Dense مع 64 وحدة وتفعيل 'relu'
classifier.add(Dense(units=27, activation='softmax'))
# إضافة طبقة Dense مع 27 وحدة وتفعيل 'softmax' لتصنيف متعدد الفئات

# تجميع الشبكة العصبية
classifier.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy']) 
# تجميع النموذج مع استخدام 'adam' كمحسن و'categorical_crossentropy' كدالة خسارة

# خطوة 2 - تحضير بيانات التدريب/الاختبار وتدريب النموذج
classifier.summary() 
# عرض ملخص للنموذج
```

```python
from keras.callbacks import ModelCheckpoint
# استيراد ModelCheckpoint لحفظ النموذج الأفضل خلال التدريب
```

```python
checkpoint = ModelCheckpoint('model-{epoch:03d}.model', monitor='val_loss', verbose=0, save_best_only=True, mode='auto')
# تهيئة ModelCheckpoint لحفظ النموذج الأفضل بناءً على 'val_loss'
history = classifier.fit(train_data, train_target, shuffle=True, epochs=20, callbacks=[checkpoint], validation_split=0.3)
# تدريب النموذج على بيانات التدريب مع تقسيم 30% للتحقق
```

```python
print(classifier.evaluate(test_data, test_target))
# تقييم النموذج على بيانات الاختبار
```

```python
N = 20  # عدد العصور (epochs) التي تم تدريب النموذج عليها
H = history  # حفظ تاريخ التدريب في متغير H
import matplotlib.pyplot as plt  # استيراد مكتبة matplotlib لرسم المخططات
plt.style.use("ggplot")  # استخدام نمط ggplot لرسم المخططات
plt.figure()  # إنشاء شكل جديد للرسم

# رسم منحنى خسارة التدريب
plt.plot(np.arange(0, N), H.history["loss"], label="train_loss")  
# رسم منحنى خسارة التحقق
plt.plot(np.arange(0, N), H.history["val_loss"], label="val_loss")
# رسم منحنى دقة التدريب
plt.plot(np.arange(0, N), H.history["accuracy"], label="train_acc")
# رسم منحنى دقة التحقق
plt.plot(np.arange(0, N), H.history["val_accuracy"], label="val_acc")

plt.title("Training Loss and Accuracy")
# إضافة عنوان للمخطط
plt.xlabel("Epoch #")
# تسمية المحور السيني
plt.ylabel("Loss/Accuracy")
# تسمية المحور الصادي
plt.legend(loc="lower left") 
# إضافة وسيلة إيضاح في الزاوية السفلية اليسرى
plt.savefig('evaluation.png') 
# حفظ المخطط في ملف صورة باسم evaluation.png
```
